```python
operators = ace.get_operators(s)
operators = operators[
    (operators['scope'] == 'REGULAR')
    & (operators['category'] == 'Time Series')
]
operators = operators[~operators['definition'].str.contains('y', na=False)].reset_index(drop=True)
time_series_ops = operators['name']
time_series_ops.to_list()
```

```python
import os
import time
import requests
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

TELEGRAM_BOT_TOKEN = os.getenv("TELEGRAM_BOT_TOKEN")
TELEGRAM_CHAT_ID = os.getenv("TELEGRAM_CHAT_ID")

def send_telegram_message(text):
    """Sends a message to the specified Telegram chat."""
    if not TELEGRAM_BOT_TOKEN or not TELEGRAM_CHAT_ID:
        print("Telegram credentials missing. Cannot send message.")
        return

    url = f"https://api.telegram.org/bot{TELEGRAM_BOT_TOKEN}/sendMessage"
    payload = {"chat_id": TELEGRAM_CHAT_ID, "text": text}
    try:
        requests.post(url, json=payload, timeout=10)
    except Exception as e:
        print(f"Failed to send telegram message: {e}")

def wait_for_telegram_command(expected_command="/continue"):
    """Pauses the script and waits for a specific command via Telegram."""
    url = f"https://api.telegram.org/bot{TELEGRAM_BOT_TOKEN}/getUpdates"
    
    # Get the latest update ID to clear out old messages before we start listening
    offset = None
    try:
        res = requests.get(url, timeout=10).json()
        if res.get("ok") and len(res["result"]) > 0:
            offset = res["result"][-1]["update_id"] + 1
    except Exception:
        pass

    prompt = f"Paused. Reply with '{expected_command}' in Telegram to request a new link and continue."
    print(prompt)
    send_telegram_message(prompt)
    
    while True:
        try:
            # Use long polling (server holds the connection open up to 'timeout' seconds)
            req_url = f"{url}?timeout=30"
            if offset:
                req_url += f"&offset={offset}"
            
            res = requests.get(req_url, timeout=35).json()
            if res.get("ok"):
                for update in res["result"]:
                    offset = update["update_id"] + 1
                    if "message" in update and "text" in update["message"]:
                        text = update["message"]["text"].strip()
                        if text == expected_command:
                            send_telegram_message(f"Command '{expected_command}' received. Resuming script...")
                            return
        except requests.exceptions.Timeout:
            continue # Timeout is normal for long-polling, just loop and try again
        except Exception as e:
            print(f"Error checking Telegram updates: {e}")
            time.sleep(5)

# --- Assuming ace, s, and alpha_list are already defined ---

batch_size = 200
all_simulation_results = []
total_simulated = 0

def authenticated():
    session = ace.check_session_timeout(s)
    if session <= 200:
        return True
    return False

# Announce startup
send_telegram_message(f"🚀 Simulation started. Total alphas to process: {len(alpha_list)}")

while alpha_list:
    batch_alphas = alpha_list[:batch_size]
    
    print(f"Simulating next {len(batch_alphas)} alphas...")
    
    # Run the simulation for the current batch
    result = ace.simulate_alpha_list_multi(s, batch_alphas)
    
    # Store the results
    all_simulation_results.append(result)
    
    # Keep track of totals
    total_simulated += len(batch_alphas)
    
    # Delete the simulated alphas from the original list
    del alpha_list[:batch_size]
    
    status_msg = f"✅ Batch complete. {total_simulated} simulated total. {len(alpha_list)} alphas remaining."
    print(status_msg)
    send_telegram_message(status_msg)
    
    if alpha_list:
        # Replaces input() with a Telegram command check
        wait_for_telegram_command(expected_command="/link")
        
        # After receiving the command in Telegram, resume processing
        ace.check_session_and_relogin(s)
        msg = "Session updated. Starting the next batch..."
        print(msg)
        send_telegram_message(msg)

final_msg = "🏁 Simulation process finished successfully."
print(final_msg)
send_telegram_message(final_msg)
```
